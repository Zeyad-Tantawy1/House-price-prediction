## Data Preprocessing

```python
# Fill missing values for categorical columns with 'None' or most frequent
categorical_cols = train.select_dtypes(include=['object']).columns

for col in categorical_cols:
    # Fill with 'None' for columns that logically can be missing
    if col in ['Alley', 'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1',
               'BsmtFinType2', 'FireplaceQu', 'GarageType', 'GarageFinish',
               'GarageQual', 'GarageCond', 'PoolQC', 'Fence', 'MiscFeature', 'MasVnrType']:
        train[col] = train[col].fillna('None')
        test[col] = test[col].fillna('None')
    else:
        # Fill with mode for other categorical columns
        mode_val = train[col].mode()[0] if len(train[col].mode()) > 0 else 'Unknown'
        train[col] = train[col].fillna(mode_val)
        test[col] = test[col].fillna(mode_val)

# Fill remaining numerical columns with median
numerical_cols = train.select_dtypes(include=np.number).columns.tolist()
numerical_cols.remove('SalePrice') # Don't fill target variable
if 'Id' in numerical_cols:
    numerical_cols.remove('Id')


for col in numerical_cols:
    if train[col].isnull().sum() > 0:
        median_val = train[col].median()
        train[col] = train[col].fillna(median_val)
    if test[col].isnull().sum() > 0:
        median_val = test[col].median()
        test[col] = test[col].fillna(median_val)

print("Missing values filled.")
```

### Subtask:
Create new features from existing ones.

```python
# Feature Engineering
def create_features(df):
    df = df.copy()

    # Total square footage
    df['TotalSF'] = df['TotalBsmtSF'] + df['1stFlrSF'] + df['2ndFlrSF']

    # Total bathrooms
    df['TotalBath'] = df['FullBath'] + df['HalfBath'] + df['BsmtFullBath'] + df['BsmtHalfBath']

    # Age of house
    df['HouseAge'] = df['YrSold'] - df['YearBuilt']
    df['RemodAge'] = df['YrSold'] - df['YearRemodAdd']

    # Garage age
    df['GarageAge'] = df['YrSold'] - df['GarageYrBlt']
    df['GarageAge'] = df['GarageAge'].fillna(0)  # No garage = 0 age

    # Total porch area
    df['TotalPorchSF'] = df['OpenPorchSF'] + df['EnclosedPorch'] + df['3SsnPorch'] + df['ScreenPorch']

    return df

train = create_features(train)
test = create_features(test)

print("New features created!")
```

### Subtask:
Encode categorical variables.

```python
# Encode categorical variables
label_encoders = {}
categorical_cols = train.select_dtypes(include=['object']).columns

for col in categorical_cols:
    le = LabelEncoder()
    # Fit on combined data to handle unseen categories
    combined_values = pd.concat([train[col], test[col]], axis=0).astype(str).unique()
    le.fit(combined_values)

    train[col] = le.transform(train[col].astype(str))
    test[col] = le.transform(test[col].astype(str))

    label_encoders[col] = le

print(f"Encoded {len(categorical_cols)} categorical columns")
```

## Feature Selection and Scaling

```python
# Select important features
important_features = [
    'TotalSF', 'GrLivArea', 'TotalBath', 'GarageCars', 'GarageArea',
    'OverallQual', 'OverallCond', 'YearBuilt', 'HouseAge',
    'Neighborhood', 'MSSubClass', 'MSZoning', 'LotArea', 'LotFrontage',
    'BsmtQual', 'KitchenQual', 'FireplaceQu', 'GarageQual',
    'TotalPorchSF', 'Foundation', 'HeatingQC'
]

# Filter features that exist in the dataset
existing_features = [f for f in important_features if f in train.columns]

X_train = train[existing_features]
y_train = train['SalePrice']
X_test = test[existing_features]

# Ensure test set has all selected features, fill missing with median if necessary (should be handled in preprocessing, but as a safeguard)
for feature in existing_features:
    if feature not in X_test.columns:
        X_test[feature] = 0 # Or a more appropriate default

# Align columns - crucial after potential feature differences
train_cols = X_train.columns
test_cols = X_test.columns

missing_in_test = set(train_cols) - set(test_cols)
for c in missing_in_test:
    X_test[c] = 0

missing_in_train = set(test_cols) - set(train_cols)
for c in missing_in_train:
    X_train[c] = 0

X_test = X_test[train_cols] # Ensure the order of columns is the same

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Features selected and scaled.")
```

## Model Training and Evaluation

```python
# Train Linear Regression model
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_scaled, y_train, test_size=0.2, random_state=42
)

model = LinearRegression()
model.fit(X_train_split, y_train_split)

y_val_pred = model.predict(X_val_split)

mse = mean_squared_error(y_val_split, y_val_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_val_split, y_val_pred)

print(f"Validation Results:")
print(f"RMSE: ${rmse:,.2f}")
print(f"R² Score: {r2:.4f}")

# Cross-validation
cv_scores = cross_val_score(model, X_train_scaled, y_train,
                            cv=5, scoring='neg_root_mean_squared_error')

print(f"\nCross-Validation Results:")
print(f"Mean RMSE: ${-cv_scores.mean():,.2f}")
print(f"Std RMSE: ${cv_scores.std():,.2f}")
```

## Final Model and Predictions

```python
# Train final model on all training data
final_model = LinearRegression()
final_model.fit(X_train_scaled, y_train)

# Make predictions on test set
test_predictions = final_model.predict(X_test_scaled)

# Create submission file
submission['SalePrice'] = test_predictions
submission.to_csv('house_prices_submission.csv', index=False)

print("Submission file created: house_prices_submission.csv")
print(f"Predicted price range: ${test_predictions.min():,.2f} - ${test_predictions.max():,.2f}")
```

## Feature Importance Analysis

```python
# Get feature importance (coefficients)
feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Coefficient': final_model.coef_
})

feature_importance['Abs_Coefficient'] = abs(feature_importance['Coefficient'])
feature_importance = feature_importance.sort_values('Abs_Coefficient', ascending=False)

print("\nTop 10 Most Important Features:")
print(feature_importance.head(10)[['Feature', 'Coefficient']])

# Plot feature importance
plt.figure(figsize=(10, 6))
top_features = feature_importance.head(10)
plt.barh(top_features['Feature'], top_features['Coefficient'])
plt.xlabel('Coefficient Value')
plt.title('Top 10 Feature Coefficients (Linear Regression)')
plt.tight_layout()
plt.show()
```

```python
# Residual plots for model evaluation
residuals = y_train - final_model.predict(X_train_scaled)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Residuals vs Fitted
axes[0].scatter(final_model.predict(X_train_scaled), residuals, alpha=0.6)
axes[0].axhline(y=0, color='red', linestyle='--')
axes[0].set_xlabel('Fitted Values')
axes[0].set_ylabel('Residuals')
axes[0].set_title('Residuals vs Fitted')

# Q-Q plot
stats.probplot(residuals, dist="norm", plot=axes[1])
axes[1].set_title('Q-Q Plot')

plt.tight_layout()
plt.show()

print("Model evaluation complete!")
```
